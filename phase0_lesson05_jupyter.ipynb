Comparing Python list comprehension and NumPy for creating 100,000 random numbers.

```python
import random
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline
```

```python
%timeit [random.random() for _ in range(100_000)]
```

```python
%timeit np.random.random(100_000)
```

```python
data = np.random.randn(1000)

plt.hist(data, bins=30)
plt.title("Random Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Count")
plt.show()
```
